```python
from Bio import Entrez

Entrez.email = "******"

handle = Entrez.esearch(
    db="pubmed",
    term="Alzheimer Disease AND (amyloid OR tau OR biomarker OR treatment)",
    retmax=100,
)

record = Entrez.read(handle)

pmids = record["IdList"]
```

```python
len(pmids)
```

```python
fetch_handle = Entrez.efetch(
    db="pubmed",
    id=",".join(pmids),
    rettype="abstract",
    retmode="xml",
)

papers = Entrez.read(fetch_handle)
```

```python
with open("data/raw/pubmed_batch.xml", "w") as f:
    f.write(papers)
```

```python
for article in papers["PubmedArticle"]:
    citation = article["MedlineCitation"]

    pmid = citation["PMID"]
    article_data = citation["Article"]
    title = article_data.get("ArticleTitle", "")

    abstract = ""

    if "Abstract" in article_data:
        abstract_sections = article_data["Abstract"]["AbstractText"]
        abstract = " ".join(str(x) for x in abstract_sections)
```
