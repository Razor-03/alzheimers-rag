```python
import json
from collections import Counter
```

```python
over_limit_by_type = Counter()
over_limit_examples = []

with open("/home/razor/Documents/alzheimers-rag/data/processed/pmc_alzheimers_children.jsonl") as f:
    for line in f:
        c = json.loads(line)
        if c["n_tokens"] > 512:
            over_limit_by_type[c["canonical_section_type"]] += 1
            if len(over_limit_examples) < 5:
                over_limit_examples.append(
                    (c["canonical_section_type"], c["n_tokens"], c["chunk_id"])
                )

print("Over-512 chunks by canonical_section_type:")
for t, count in over_limit_by_type.most_common():
    print(f"  {t}: {count}")

print("\nSample over-limit chunks:")
for t, n, cid in over_limit_examples:
    print(f"  [{t}] {n} tokens -- {cid}")
```

```python
with open("/home/razor/Documents/alzheimers-rag/data/processed/pmc_alzheimers_children.jsonl") as f:
    children = [json.loads(l) for l in f]

overflow_body = [
    c for c in children
    if c["n_tokens"] > 512
    and c["canonical_section_type"] in {"introduction", "figure_footnote"}
    # and c["canonical_section_type"] in {"table_footnote"}
]

for c in overflow_body[:50]:
    print(f"--- {c['chunk_id']}  ({c['n_tokens']} tokens) - {c['canonical_section_type']} ---")
    print(c["text"][:])
    print("...")
    print()
```
